```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
nadia = cv2.imread("../DATA/Nadia_Murad.jpg", 0)
dennis = cv2.imread("../DATA/Denis_Mukwege.jpg", 0)
solvay = cv2.imread("../DATA/solvay_conference.jpg", 0)

```

```python
faceCascade = cv2.CascadeClassifier("../DATA/haarcascades/haarcascade_frontalface_default.xml")
```

```python
def detect_face(img):
    face_image = img.copy()
    face_rects = faceCascade.detectMultiScale(face_image)
    for (x,y, w, h) in face_rects:
        cv2.rectangle(face_image,(x,y), (x+w, y+h), (255, 255, 255), thickness=10)
    return face_image


        
```

```python
result = detect_face(dennis)
plt.imshow(result, cmap="gray")
```

```python
result = detect_face(nadia)
plt.imshow(result, cmap="gray")
```

```python
result = detect_face(solvay)
plt.imshow(result, cmap="gray")
```

```python
def adj_detect_face(img):
    face_image = img.copy()
    face_rects = faceCascade.detectMultiScale(face_image, scaleFactor=1.2, minNeighbors=5)
    for (x,y,w,h) in face_rects:
        cv2.rectangle(face_image, (x,y), (x+w, y+h), (255, 255, 255), 10)
    return face_image


```

```python
result = adj_detect_face(solvay)
plt.imshow(result, cmap="gray")
```

```python
eye_cascade = cv2.CascadeClassifier("../DATA/haarcascades/haarcascade_eye.xml")
def detect_eye(img):
    face_image = img.copy()
    eye_rects = eye_cascade.detectMultiScale(face_image)
    for (x,y, w, h) in eye_rects:
        cv2.rectangle(face_image,(x,y), (x+w, y+h), (255, 255, 255), thickness=10)
    return face_image
```

```python
result = detect_eye(nadia)
plt.imshow(result, cmap="gray")
```

```python
# no rects because of color alterations in this image.
result = detect_eye(dennis)
plt.imshow(result, cmap="gray")
```

```python
cap = cv2.VideoCapture(0)
while True:
    ret, frame = cap.read(0)
    frame = detect_face(frame)
    cv2.imshow("video face detect", frame)
    k = cv2.waitKey(1)
    if k == 27:
        break
```
